# Hospital Readmission Analytics: Data Cleaning & Preprocessing Notebook
**Dataset:** UCI Diabetes 130-US Hospitals (1999–2008)
**Objective:** Clean raw electronic health records (EHR) data, remove non-eligible encounters (deceased/hospice), engineer analytical features, deduplicate multi-visit patients, and export standardized datasets for SQL and Tableau analysis.

```python
import pandas as pd
import numpy as np
import os

# Set relative file paths
RAW_DATA_PATH = '../data/raw/diabetic_data.csv'
CLEAN_ENCOUNTERS_PATH = '../data/cleaned_diabetic_data.csv'
CLEAN_PATIENTS_PATH = '../data/cleaned_patient_first_visits.csv'

print("Libraries loaded successfully.")
```

## 1. Load Raw Dataset & Inspect Initial Records

```python
df_raw = pd.read_csv(RAW_DATA_PATH)
print(f"Initial Raw Encounter Records: {len(df_raw):,}")
print(f"Initial Features Count: {df_raw.shape[1]}")
df_raw.head(3)
```

## 2. Identify & Handle Missing Values (`?` Strings)
In this dataset, missing values are encoded as `'?'`. We convert them to standard `NaN` and drop columns with prohibitive missingness (>35%).

```python
# Replace '?' with NaN
df = df_raw.replace('?', np.nan)

# Calculate missingness percentages
missing_summary = df.isnull().sum()
missing_summary = missing_summary[missing_summary > 0].sort_values(ascending=False)
missing_df = pd.DataFrame({
    'Missing Count': missing_summary,
    'Percentage (%)': (missing_summary / len(df)) * 100
})
missing_df
```

```python
# Drop weight (~97% missing), payer_code (~40% missing), and medical_specialty (~49% missing)
cols_to_drop = ['weight', 'payer_code', 'medical_specialty']
df_clean = df.drop(columns=cols_to_drop)
print(f"Remaining columns after dropping high-missing features: {df_clean.shape[1]}")
```

## 3. Filter Non-Eligible Encounters (Deceased & Hospice Patients)
Patients discharged to Hospice (`13`, `14`) or Expired/Deceased (`11`, `19`, `20`, `21`) cannot be readmitted within 30 days. Retaining them biases readmission risk downward.

```python
EXPIRED_HOSPICE_IDS = [11, 13, 14, 19, 20, 21]
expired_mask = df_clean['discharge_disposition_id'].isin(EXPIRED_HOSPICE_IDS)
removed_expired_count = expired_mask.sum()

df_valid = df_clean[~expired_mask].copy()
print(f"Encounters removed (Deceased / Hospice): {removed_expired_count:,}")
print(f"Valid post-discharge encounters: {len(df_valid):,}")
```

## 4. Target Variable Standardization & Feature Bucketing
Create binary target variable `readmitted_30day` (1 = readmitted in <30 days, 0 = otherwise) and categorical buckets for SQL and Tableau reporting.

```python
df_valid['readmitted_30day'] = np.where(df_valid['readmitted'] == '<30', 1, 0)

# Age standardization
df_valid['age_clean'] = df_valid['age'].str.replace('[', '', regex=False).str.replace(')', '', regex=False)

# Prior Inpatient Bucket
df_valid['prior_inpatient_bucket'] = df_valid['number_inpatient'].apply(
    lambda x: '0 Visits' if x == 0 else ('1 Visit' if x == 1 else ('2 Visits' if x == 2 else '3+ Visits'))
)

print("Readmission 30-day distribution in valid encounters:")
    print(df_valid['readmitted_30day'].value_counts(normalize=True) * 100)
```

## 5. Deduplicate Multi-Visit Patients
To eliminate potential independence violations when analyzing patient-level factors, we also isolate each patient's index (first) hospital visit.

```python
df_sorted = df_valid.sort_values(['patient_nbr', 'encounter_id'])
df_first_visit = df_sorted.drop_duplicates(subset=['patient_nbr'], keep='first').copy()

print(f"Total Valid Encounters: {len(df_valid):,}")
print(f"Unique Patients (First Visit): {len(df_first_visit):,}")
print(f"Secondary Duplicate Encounters Dropped: {len(df_valid) - len(df_first_visit):,}")
```

## 6. Audit Summary & Export Data

```python
audit_df = pd.DataFrame({
    'Step': [
        '1. Raw Dataset Loaded',
        '2. Excluded Deceased/Hospice Encounters',
        '3. Deduplicated Secondary Patient Visits'
    ],
    'Records Remaining': [
        len(df_raw),
        len(df_valid),
        len(df_first_visit)
    ],
    'Records Removed': [
        0,
        removed_expired_count,
        len(df_valid) - len(df_first_visit)
    ]
})
print(audit_df.to_string(index=False))

# Save datasets
df_valid.to_csv(CLEAN_ENCOUNTERS_PATH, index=False)
df_first_visit.to_csv(CLEAN_PATIENTS_PATH, index=False)
print("Databases saved successfully.")
```
